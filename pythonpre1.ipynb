```python
# gusseing game
# generate a random integer between 1 and 100

import random
jackpot = random.randint(1,100)
guess = int(input('Please enter jackpot number'))
counter = 1
while guess != jackpot:
  if guess < jackpot:
    print('Wrong!! Please enter higher number')
  else:
    print('Wrong!! Please enter lower number')
  guess = int(input('Please enter jackpot number'))
  counter = counter + 1
else:
  print('Congratulation!! You are correct at ',counter,' round')
```

```python
"""Create a simple electricity bill calculator:

0–100 units → ₹5/unit
101–200 → ₹7/unit
Above 200 → ₹10/unit
  """
x=int(float(input("enter unt to be used")))
if 0<x<=100:
    bill=x*5
elif 100<x<=200:
    bill=x*7
else:
    bill=x*10
t=(bill+150)
print(f"TOTALL BILL:",t)       
```

```python
# Find even numbers from 1 to 20
evens = []
for i in range(1, 21):
    if i % 2 == 0:   # check divisibility by 2
        evens.append(i)

print(evens)
```

```python
# Find even numbers from 1 to 20
evens = []
for i in range(1, 21):
    if i % 2 != 0:   # check divisibility by 2
        evens.append(i)

print(evens)
```

```python
evens = [i for i in range(1, 21) if i % 2 == 0]
print(evens)
```
