```python
def first_missing_positive(nums):
    nums = set(nums)
    n = 1

    while n in nums:
        n += 1

    return n

print(first_missing_positive([3,7,-1,2,5,1]))
```

```python
def first_missing_possite(num):
    num=set(num)
    while n in num:
        n+=1
    return n
```

```python

def facq(L):
    L = set(L)
    i = 1

    while i in L:
        i += 1

    return i

print(facq([3, 4, -1, 2, 5, 1]))
```

```python
nums = [3, 1, 2]
len(nums)                    # 3
sorted(nums)                 # [1, 2, 3], returns a new list
sorted(nums, reverse=True)   # [3, 2, 1]
reversed(nums)               # iterator, wrap with list()
enumerate(nums, start=1)     # (1, 3), (2, 1), (3, 2)
zip([1, 2], ["a", "b"])      # (1, 'a'), (2, 'b')
range(5)                     # 0, 1, 2, 3, 4
```
